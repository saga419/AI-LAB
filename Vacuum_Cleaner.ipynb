```python
class Environment:

    def __init__(self):
        self.locations = {"A": "Dirty", "B": "Dirty"}
        self.agent_location = "A"

    def is_clean(self):
        return all(status == "Clean" for status in self.locations.values())


class ReflexVacuumAgent:

    def act(self, location, status):
        if status == "Dirty":
            return "Suck"
        elif location == "A":
            return "Right"
        elif location == "B":
            return "Left"


def run_simulation():
    env = Environment()
    agent = ReflexVacuumAgent()
    step = 1

    print(f"Initial State: {env.locations}, Vacuum at: {env.agent_location}\n")

    while not env.is_clean() and step <= 10:
        current_loc = env.agent_location
        current_status = env.locations[current_loc]

        action = agent.act(current_loc, current_status)
        print(f"Step {step}: Location = {current_loc} [{current_status}] -> Action = {action}")

        if action == "Suck":
            env.locations[current_loc] = "Clean"
        elif action == "Right":
            env.agent_location = "B"
        elif action == "Left":
            env.agent_location = "A"

        step += 1

    print(f"\nFinal State: {env.locations} (All Clean!)")


if __name__ == "__main__":
    run_simulation()
```
