```python

from itertools import combinations
import random

def cost(state):
    count = 0
    for i, j in combinations(range(len(state)), 2):
        if abs(state[i] - state[j]) == abs(i - j):
            count += 1
    return count

def hill_climbing(n):
    state = list(range(n))
    random.shuffle(state)
    step = 1

    while True:
        current_cost = cost(state)
        print("\nStep:", step)
        print("State:", state)
        print("Cost:", current_cost)

        if current_cost == 0:
            break

        neighbors = []

        for i, j in combinations(range(n), 2):
            new_state = state.copy()
            new_state[i], new_state[j] = new_state[j], new_state[i]
            neighbors.append((cost(new_state), i, j, new_state))

        best = min(neighbors, key=lambda x: (x[0], x[1], x[2]))

        if best[0] >= current_cost:
            print("Local minimum reached. Restarting...")
            state = list(range(n))
            random.shuffle(state)
            step = 1
            continue

        state = best[3]
        step += 1

    print("\nSolution:")
    for row in range(n):
        print(" ".join("Q" if state[col] == row else "."
                       for col in range(n)))

n = int(input("Enter number of queens: "))
hill_climbing(n)
```

```python

from itertools import product

def implies(a, b):
    return not a or b

print("P Q R | Q->P P->~Q Q|R | KB")

models = []

for P, Q, R in product([False, True], repeat=3):
    s1 = implies(Q, P)
    s2 = implies(P, not Q)
    s3 = Q or R
    KB = s1 and s2 and s3

    print(
        f"{int(P)} {int(Q)} {int(R)} |"
        f"  {int(s1)}     {int(s2)}    {int(s3)} |"
        f" {int(KB)}"
    )

    if KB:
        models.append((P, Q, R))

queries = {
    "KB entails R": lambda P, Q, R: R,
    "KB entails R -> P": lambda P, Q, R: implies(R, P),
    "KB entails Q -> R": lambda P, Q, R: implies(Q, R)
}

print("\nResults:")
for query, evaluate in queries.items():
    result = all(evaluate(*m) for m in models)
    print(query, ":", result)

print("\nKB models:")
for model in models:
    print(model)
```
