```python

# Knowledge Base using First Order Logic
# Proving Query using Forward Chaining

# Initial Knowledge Base (Facts)
KB = {
    ("Parent", "John", "Mary"),
    ("Parent", "Mary", "Alice")
}

# Query to be proved
query = ("Grandparent", "John", "Alice")

# Forward Chaining Algorithm
def forward_chaining(KB):

    inferred = True

    while inferred:
        inferred = False
        new_facts = set()

        # Rule 1: Parent(x,y) -> Human(x)
        for fact in KB:
            if fact[0] == "Parent":
                x = fact[1]
                conclusion = ("Human", x)

                if conclusion not in KB:
                    new_facts.add(conclusion)

        # Rule 2: Human(x) -> Mortal(x)
        for fact in KB:
            if fact[0] == "Human":
                x = fact[1]
                conclusion = ("Mortal", x)

                if conclusion not in KB:
                    new_facts.add(conclusion)

        # Rule 3:
        # Parent(x,y) AND Parent(y,z)
        # -> Grandparent(x,z)

        for fact1 in KB:
            if fact1[0] == "Parent":

                x = fact1[1]
                y = fact1[2]

                for fact2 in KB:
                    if fact2[0] == "Parent":

                        y2 = fact2[1]
                        z = fact2[2]

                        if y == y2:
                            conclusion = (
                                "Grandparent", x, z
                            )

                            if conclusion not in KB:
                                new_facts.add(conclusion)

        # Add newly inferred facts
        if new_facts:
            KB.update(new_facts)
            inferred = True

            print("\nNew Facts Inferred:")
            for fact in new_facts:
                print(fact)

    return KB


# Execute Forward Chaining
result = forward_chaining(KB)

# Check Query
print("\nQuery:", query)

if query in result:
    print("Query Proved: TRUE")
else:
    print("Query Cannot Be Proved: FALSE")

# Display Final Knowledge Base
print("\nFinal Knowledge Base:")
for fact in sorted(result):
    print(fact)
```

```python

# Implement Unification in First Order Logic
# Both Unified and Non-Unified Cases

def is_variable(x):
    return isinstance(x, str) and x.startswith("?")

def apply_substitution(x, substitution):
    while is_variable(x) and x in substitution:
        x = substitution[x]
    return x

def unify(x, y, substitution):

    x = apply_substitution(x, substitution)
    y = apply_substitution(y, substitution)

    # Case 1: Both expressions are identical
    if x == y:
        return substitution

    # Case 2: x is a variable
    if is_variable(x):
        substitution[x] = y
        return substitution

    # Case 3: y is a variable
    if is_variable(y):
        substitution[y] = x
        return substitution

    # Case 4: Different constants
    return None


def unify_expressions(E1, E2):

    # Predicate symbols must match
    if E1[0] != E2[0]:
        return None

    # Number of arguments must match
    if len(E1) != len(E2):
        return None

    substitution = {}

    # Unify each argument
    for a, b in zip(E1[1:], E2[1:]):
        substitution = unify(a, b, substitution)

        if substitution is None:
            return None

    return substitution


# -------------------------------
# CASE 1: SUCCESSFUL UNIFICATION
# -------------------------------

E1 = ("Eats", "?x", "Apple")
E2 = ("Eats", "Sagar", "?y")

print("CASE 1: SUCCESSFUL UNIFICATION")
print("Expression 1:", E1)
print("Expression 2:", E2)

result = unify_expressions(E1, E2)

if result is not None:
    print("Unification Successful")
    print("Substitution Set:", result)

    unified = tuple(
        apply_substitution(item, result)
        for item in E1
    )

    print("Unified Expression:", unified)

else:
    print("Unification Failed")


# -------------------------------
# CASE 2: FAILED UNIFICATION
# -------------------------------

E1 = ("Knows", "Sagar", "?x")
E2 = ("Knows", "?x", "Elizabeth")

print("\nCASE 2: FAILED UNIFICATION")
print("Expression 1:", E1)
print("Expression 2:", E2)

result = unify_expressions(E1, E2)

if result is not None:
    print("Unification Successful")
    print("Substitution Set:", result)

else:
    print("Unification Failed")
```
