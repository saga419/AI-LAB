```python
import math


class TicTacToe:

    def __init__(self, human_symbol, ai_symbol):
        self.board = [" " for _ in range(9)]
        self.human = human_symbol
        self.ai = ai_symbol

    def print_board(self):
        for row in [self.board[i * 3 : (i + 1) * 3] for i in range(3)]:
            print("| " + " | ".join(row) + " |")

    def print_board_indices(self):
        number_board = [[str(i) for i in range(j * 3, (j + 1) * 3)] for j in range(3)]
        for row in number_board:
            print("| " + " | ".join(row) + " |")

    def available_moves(self):
        return [i for i, spot in enumerate(self.board) if spot == " "]

    def empty_squares(self):
        return " " in self.board

    def make_move(self, square, letter):
        if self.board[square] == " ":
            self.board[square] = letter
            return True
        return False

    def check_winner(self, board, letter):
        win_conditions = [
            [0, 1, 2],
            [3, 4, 5],
            [6, 7, 8],
            [0, 3, 6],
            [1, 4, 7],
            [2, 5, 8],
            [0, 4, 8],
            [2, 4, 6],
        ]
        for condition in win_conditions:
            if all(board[i] == letter for i in condition):
                return True
        return False


def minimax(state, depth, is_maximizing, ai, human):
    if state.check_winner(state.board, ai):
        return {"position": None, "score": 10 - depth}
    elif state.check_winner(state.board, human):
        return {"position": None, "score": -10 + depth}
    elif not state.empty_squares():
        return {"position": None, "score": 0}

    if is_maximizing:
        best = {"position": None, "score": -math.inf}
        for possible_move in state.available_moves():
            state.board[possible_move] = ai
            sim_score = minimax(state, depth + 1, False, ai, human)

            state.board[possible_move] = " "
            sim_score["position"] = possible_move

            if sim_score["score"] > best["score"]:
                best = sim_score
        return best
    else:
        best = {"position": None, "score": math.inf}
        for possible_move in state.available_moves():
            state.board[possible_move] = human
            sim_score = minimax(state, depth + 1, True, ai, human)

            state.board[possible_move] = " "
            sim_score["position"] = possible_move

            if sim_score["score"] < best["score"]:
                best = sim_score
        return best


def play_game():
    print("Welcome to Tic-Tac-Toe!")

    human_symbol = ""
    while human_symbol not in ["X", "O"]:
        human_symbol = input("Choose your symbol ('X' or 'O'): ").strip().upper()

    ai_symbol = "O" if human_symbol == "X" else "X"

    first_player = ""
    while first_player not in ["HUMAN", "COMPUTER"]:
        first_player = (
            input("Who goes first? Type 'HUMAN' or 'COMPUTER': ").strip().upper()
        )

    game = TicTacToe(human_symbol, ai_symbol)

    print("\nBoard position mapping:")
    game.print_board_indices()
    print("\nStarting game...\n")

    turn = "HUMAN" if first_player == "HUMAN" else "COMPUTER"

    while game.empty_squares():
        if turn == "HUMAN":
            valid_move = False
            while not valid_move:
                try:
                    square = int(input(f"Your move ({game.human}) [0-8]: "))
                    if square in game.available_moves():
                        game.make_move(square, game.human)
                        valid_move = True
                    else:
                        print("Invalid square or already taken. Try again.")
                except ValueError:
                    print("Please enter a number between 0 and 8.")
        else:
            print(f"COMPUTER ({game.ai}) is making a move...")
            move = minimax(game, 0, True, game.ai, game.human)["position"]
            game.make_move(move, game.ai)

        game.print_board()
        print("\n")

        current_symbol = game.human if turn == "HUMAN" else game.ai
        if game.check_winner(game.board, current_symbol):
            print(f"{'HUMAN wins!' if turn == 'HUMAN' else 'COMPUTER wins!'}")
            return

        turn = "COMPUTER" if turn == "HUMAN" else "HUMAN"

    print("The game ends in a draw!")


if __name__ == "__main__":
    play_game()
```
