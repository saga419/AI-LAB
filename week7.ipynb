```python

# Alpha-Beta Pruning Algorithm

# Function to display pruned nodes
def show_pruned(node, path):
    if isinstance(node, int):
        print(f"    Pruned leaf {path} = {node}")
        return

    name, children = node
    print(f"    Pruned subtree: {path}->{name}")

    for i, child in enumerate(children):
        show_pruned(child, f"{path}->{name}[{i+1}]")


# Alpha-Beta Pruning function
def alpha_beta(node, alpha, beta, maximizing, path="Root"):

    # If node is a leaf
    if isinstance(node, int):
        print(f"Visited {path} = {node}")
        return node

    name, children = node

    if maximizing:
        value = float('-inf')

        for i, child in enumerate(children):
            child_path = f"{path}->{name}[{i+1}]"

            value = max(
                value,
                alpha_beta(child, alpha, beta, False, child_path)
            )

            alpha = max(alpha, value)

            # Pruning condition
            if alpha >= beta:
                print(f"Pruning at {name}")
                for j in range(i + 1, len(children)):
                    show_pruned(
                        children[j],
                        f"{path}->{name}[{j+1}]"
                    )
                break

        print(f"{name} (MAX) = {value}")
        return value

    else:
        value = float('inf')

        for i, child in enumerate(children):
            child_path = f"{path}->{name}[{i+1}]"

            value = min(
                value,
                alpha_beta(child, alpha, beta, True, child_path)
            )

            beta = min(beta, value)

            # Pruning condition
            if alpha >= beta:
                print(f"Pruning at {name}")
                for j in range(i + 1, len(children)):
                    show_pruned(
                        children[j],
                        f"{path}->{name}[{j+1}]"
                    )
                break

        print(f"{name} (MIN) = {value}")
        return value


# Given game tree
tree = (
    "Root",
    [
        ("A", [
            ("B", [10, 9]),
            ("C", [14, 18])
        ]),
        ("D", [
            ("E", [5, 4]),
            ("F", [50, 3])
        ])
    ]
)

# Start Alpha-Beta Pruning
print("Alpha-Beta Pruning")
print("------------------")

result = alpha_beta(
    tree,
    float('-inf'),
    float('inf'),
    True
)

print("\nOptimal Root Value =", result)
```

```python

import random
import math

N = 8

# Calculate number of attacking queen pairs
def calculate_cost(state):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):

            # Same row
            if state[i] == state[j]:
                conflicts += 1

            # Same diagonal
            elif abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


# Display chessboard
def display_board(state):
    print("\nChessboard:")

    for row in range(N):
        for col in range(N):
            if state[col] == row:
                print(" Q ", end="")
            else:
                print(" . ", end="")
        print()


# Simulated Annealing
def simulated_annealing():

    current = [random.randint(0, N - 1) for _ in range(N)]
    current_cost = calculate_cost(current)

    temperature = 10.0
    cooling_rate = 0.9995

    iteration = 0

    print("Initial State:", current)
    print("Initial Conflicts:", current_cost)

    while current_cost != 0 and iteration < 10000:

        # Select a random column
        col = random.randint(0, N - 1)

        # Select a random new row
        new_row = random.randint(0, N - 1)

        next_state = current.copy()
        next_state[col] = new_row

        next_cost = calculate_cost(next_state)

        delta = current_cost - next_cost

        # Accept better solution
        if delta > 0:
            current = next_state
            current_cost = next_cost

        # Accept worse solution probabilistically
        else:
            probability = math.exp(delta / temperature)

            if random.random() < probability:
                current = next_state
                current_cost = next_cost

        # Cool the temperature
        temperature *= cooling_rate
        iteration += 1

    print("\nFinal State:", current)
    print("Final Conflicts:", current_cost)
    print("Iterations:", iteration)

    display_board(current)

    if current_cost == 0:
        print("\nSolution Found!")
    else:
        print("\nSolution not found within iteration limit.")


# Run the algorithm
random.seed(42)
simulated_annealing()
```
