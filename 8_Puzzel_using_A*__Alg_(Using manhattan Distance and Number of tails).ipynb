```python
import heapq

# Goal state
GOAL = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

#


# ==================================================
# HEURISTIC 1: NUMBER OF TILES OUT OF PLACE
# ==================================================
def misplaced_tiles(state):
    count = 0

    for i in range(9):
        if state[i] != 0 and state[i] != GOAL[i]:
            count += 1

    return count


# ==================================================
# HEURISTIC 2: MANHATTAN DISTANCE
# ==================================================
def manhattan_distance(state):
    distance = 0

    for i in range(9):
        tile = state[i]

        if tile != 0:

            # Current position
            row = i // 3
            col = i % 3

            # Goal position
            goal_pos = GOAL.index(tile)
            goal_row = goal_pos // 3
            goal_col = goal_pos % 3

            distance += abs(row - goal_row)
            distance += abs(col - goal_col)

    return distance


# ==================================================
# GENERATE NEIGHBOURS
# ==================================================
def get_neighbors(state):

    neighbors = []

    zero = state.index(0)

    row = zero // 3
    col = zero % 3

    moves = [
        (-1, 0),   # Up
        (1, 0),    # Down
        (0, -1),   # Left
        (0, 1)     # Right
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_zero = new_row * 3 + new_col

            new_state = list(state)

            # Swap blank with tile
            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            neighbors.append(tuple(new_state))

    return neighbors


# ==================================================
# A* ALGORITHM
# ==================================================
def a_star(start, heuristic):

    # Priority queue
    # (f, g, state)
    open_list = []

    g_cost = {start: 0}

    parent = {start: None}

    h = heuristic(start)

    heapq.heappush(open_list, (h, 0, start))

    closed = set()

    while open_list:

        f, g, current = heapq.heappop(open_list)

        # Goal reached
        if current == GOAL:

            path = []

            while current is not None:
                path.append(current)
                current = parent[current]

            path.reverse()

            return path

        if current in closed:
            continue

        closed.add(current)

        # Generate neighbours
        for neighbor in get_neighbors(current):

            if neighbor in closed:
                continue

            new_g = g + 1

            if neighbor not in g_cost or new_g < g_cost[neighbor]:

                g_cost[neighbor] = new_g

                parent[neighbor] = current

                h = heuristic(neighbor)

                f = new_g + h

                heapq.heappush(
                    open_list,
                    (f, new_g, neighbor)
                )

    return None


# ==================================================
# PRINT PUZZLE
# ==================================================
def print_puzzle(state):

    for i in range(0, 9, 3):
        print(state[i:i+3])

    print()


# ==================================================
# INITIAL STATE
# ==================================================

start = ( 1, 2, 3,
         4, 0, 6,
          7, 5, 8
        )


# ==================================================
# A* WITH NUMBER OF TILES OUT OF PLACE
# ==================================================

print("==========================================")
print("A* - NUMBER OF TILES OUT OF PLACE")
print("==========================================")

print("Initial State:")
print_puzzle(start)

print("Heuristic Value:",
      misplaced_tiles(start))

solution1 = a_star(start, misplaced_tiles)

if solution1:

    print("Solution Found!")
    print("Number of Moves:",
          len(solution1) - 1)
    print()

    for i, state in enumerate(solution1):

        print("Step", i)
        print_puzzle(state)

else:
    print("No solution exists.")


# ==================================================
# A* WITH MANHATTAN DISTANCE
# ==================================================

print("==========================================")
print("A* - MANHATTAN DISTANCE")
print("==========================================")

print("Initial State:")
print_puzzle(start)

print("Heuristic Value:",
      manhattan_distance(start))

solution2 = a_star(start, manhattan_distance)

if solution2:

    print("Solution Found!")
    print("Number of Moves:",
          len(solution2) - 1)
    print()

    for i, state in enumerate(solution2):

        print("Step", i)
        print_puzzle(state)

else:
    print("No solution exists.")
```
